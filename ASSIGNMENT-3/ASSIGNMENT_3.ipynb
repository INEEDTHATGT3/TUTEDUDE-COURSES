## Task 1 - Basic Function: Price After Discount
 Write a function apply_discount(price, discount_percent) that: 
1. Returns the price after discount.
2. If discount_percent is missing, apply a default discount of 5%.
3. Test the function with: (i) apply_discount(1000, 10) ,(ii) apply_discount(500) # uses default discount 
4. Extra (optional): Add a condition inside the function to ensure discount never exceeds 60%.

```python
def apply_discount(price, discount_percent=5):
    # 60% capped
    if discount_percent>60:
        discount_percent=60

    # Calculation : Price * (1-Discount/100) 
    final_price = price * (1- discount_percent/100)
    return round(final_price, 2)

# Test case
print(f"Test 1: (1000, 10): {apply_discount(1000, 10)}")  
print(f"Test 1: (500): {apply_discount(500)}")
print(f"Test 1: (1000, 75): {apply_discount(1000, 75)}") 

# upon reading recommendations of assignment 2 doing changes
# function -> if check -> variable formula used -> output
```

## Task 2 - Recursive Function: Factorial Utility
- Create a recursive function factorial(n) that:
1. Returns the factorial of n.
2. Handles two edge cases: n == 0 and n == 1.
3. Prints an error message if n is negative.
- Test with:
1. factorial(5)
2. factorial(0)
3. factorial(-3)

```python
def factorial(n):
    # invalid input
    if n<0:
        print(f"Invalid input: {n}, is negative")
        return None     
    # base case
    if n == 0 or n == 1:
       return 1
        
    # recursive case - ek case hum sambhal lege baaki recursion
    return n * factorial(n-1)

# Test Case
print(f"Factorial(5): {factorial(5)}")
print(f"Factorial(0): {factorial(0)}")
print(f"Factorial(-3): {factorial(-3)}")
```

## Task 3 - Lambda Function: GST Calculator
- Create a lambda function gst that returns price after adding 18% GST.
Example:
1. gst = lambda price: price + (0.18 * price)
2. print(gst(100)) # should return 118
- Extra (optional): Create another lambda to compute final price after GST + discount together.

```python
# Lambda for 18% GST
gst = lambda price: price + (0.18 * price)

# Extra: lambda for Final price After GST and Discount
final_price_calc = lambda price, discount: gst(price) * (1-discount/100)

# Test case
base_price = 1000
discount_val = 10

# Testing
print(f"Price after GST: {gst(base_price): .2f}")
print(f"Final Price after GST and {discount_val}% Discount: {final_price_calc(base_price, discount_val): .2f}")
```

## Task 4 - Using map(): Apply GST to List of Prices
- Given: prices = [100, 250, 400, 1200, 50]
1. Use map() with your gst lambda to generate a new list prices_with_gst.
2. Print both lists: Original prices , Prices after GST

```python
# GST Lambda
gst = lambda price: round(price * 1.18, 2)

# List of prices
prices = [100, 250, 400, 1200, 50]

# Using Map() 
prices_with_gst = list(map(gst, prices))

# Vertical comparison print -> better visual appeal
print(f"{'Original Price':<20} | {'Price after GST':<20}")
print("-" * 35)
for orig, taxed in zip(prices, prices_with_gst):
    print(f" INR {orig:<15} | INR {taxed: <15.2f}")

```

## Task 5 - Using filter(): Filter Expensive Products
- Given the list: prices = [100, 250, 400, 1200, 50, 2000, 850]
- Use filter() to:
1. Create a list of prices greater than 500.
2. Create another list of prices less than or equal to 500.
3. Print both lists.


```python
def is_expensive(price):
    return price>500

# Input Data
prices = [100, 250, 400, 1200, 50, 2000, 850]

#1. Filter prices > 500 and sort them
expensive_prices = sorted(list(filter(is_expensive, prices)))

#2. Filter prices =<500 and sort them
budget_prices = sorted(list(filter(lambda x: not is_expensive(x), prices)))

# Output results
print(f"Expensive (>500): {expensive_prices}")
print(f"Budget (=<500): {budget_prices} " )
```

## Task 6 - Combined Utility Function
- Write a function process_prices(prices) that:
1. Takes a list of prices.
2. Uses map() + lambda to apply 10% discount to all prices.
3. Uses filter() to keep only discounted prices that are above 300.
4. Returns both lists:
- discounted_prices
- filtered_prices
- Test with: process_prices[500, 900, 50, 750]

```python
def process_prices(prices):
    # Exception handiling
    if not prices:
        return "No prices provided"
    # 1. Applying 10% discount and rounding to 2% decimal places
    discounted_prices = list(map(lambda p: round(p * 0.9, 2), prices))

    #2. Filtering  out prices
    filtered_prices = list(filter(lambda p: p>300, discounted_prices))

    return{
        "discounted_prices" : discounted_prices,
        "filtered prices" : filtered_prices, # returned dict -> jyada readability
    }

# Test case 
test_prices = [500, 900, 50, 750]
results = process_prices(test_prices)

print(f"Orignal Input: {test_prices}")
print(f"All dicounted prices: {results['discounted_prices']}")
print(f"Filtered prices (>300): {results['filtered prices']}")

process_prices([])
```

## Task 7 - Mini Problem: Menu Using Functions
- Create the following three small functions:
1. add_price(prices_list, price) —+ adds price to the list.
2. get_average_price(prices list) —+ returns average price.
3. get_max_price(prices_list) -+ returns the maximum price.
- Then create a simple menu:
1. —+ Add price
2. —+ Show average price
3. —+ Show highest price
4. q —+ Quit
- Use only loops + function calls (no OOP).

```python
# functions defined
def add_price(prices_list, price):
    prices_list.append(price)

def get_average_price(prices_list):
    if not prices_list:
        return 0
    return sum(prices_list)/ len(prices_list)

def get_max_price(prices_list):
    if not prices_list:
        return 0
    return max(prices_list)

inventory_prices = []

# looping conditions
while True:
    print("\n Biling Utility Menu")
    print("1 -> Add price")
    print("2 -> Show average price")
    print("3 -> Show highest price")
    print("q -> Quit")

    choice = input("Select an opton: ").strip().lower()

    if choice == '1':
        raw_input = input("Enter the price to add: ")
        #input validation without creating exception (tried)
        if raw_input.replace('.', '', 1).isdigit():
            new_price = float(raw_input)
            add_price(inventory_prices, new_price)
            print(f"Price INR {new_price:.2f} added successfully.")
        else:
            print("invalid input. Please enter a valid number")
    
    elif choice == '2':
        avg = get_average_price(inventory_prices)
        print(f"Average Price: INR {avg:.2f}")

    elif choice == '3':
        highest = get_max_price(inventory_prices)
        print(f"Highest Price: INR {highest:.2f}")

    elif choice == 'q':
        print("Exiting billing utility. Goodbye!")
        break

    else:
        print("Invalid choice. Please select 1, 2, 3, or q.")
```
