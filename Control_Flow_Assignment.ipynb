### Task 1: Discount Rules (if I elif I else)
1. Write a program that reads an integer order_amount from the user using input().
2. Apply the following discount rules and print the final amount:
2.1 order amount >= 2000 - 15% discount
2.2 1500 <= order amount < 2000 - 10% discount
2.3 1000 <= order amount < 1500 - 7% discount
2.4 otherwise -+ 0% discount
3. Ensure you convert input to a number and handle the case when the input is non-numeric by displaying an error message and exiting the program.
4. Extra (optional): Add tax (fixed 5%) after discount and print subtotal, tax, and final total (still within conditionals and basic arithmetic).

```python
# Task 1: Discount Rules

Order_Amount = input("Enter the order amount: ")

# 2. Handling non-numeric 
if Order_Amount.isdigit():
    # Convert input to an integer
    order_amount = int(Order_Amount)
    
    # 2. Apply discount rules using if / elif / else
    if order_amount >= 2000:
        discount_percent = 15
    elif 1500 <= order_amount < 2000:
        discount_percent = 10
    elif 1000 <= order_amount < 1500:
        discount_percent = 7
    else:
        discount_percent = 0

    # Basic arithmetic for discount
    discount_amount = (discount_percent / 100) * order_amount
    subtotal = order_amount - discount_amount

    # Adding tax (fixed 5%) and print breakdown
    tax_rate = 0.05
    tax_amount = subtotal * tax_rate
    final_total = subtotal + tax_amount

    # Final Output
    print("-" * 20)
    print(f"Order Amount: INR {order_amount}")
    print(f"Discount: {discount_percent}% (INR {discount_amount})")
    print(f"Subtotal: INR {subtotal}")
    print(f"Tax (5%): INR {tax_amount}")
    print(f"Final Total: INR {final_total}")
    print("-" * 20)

else:
    # Handle the case when the input is non-numeric
    print("Error: Invalid input. Please enter a numeric integer value.")
```

### Task 2: Process Multiple Orders (for loop)
1. Given a list of order amounts: orders = [1200, 2500, 800, 1750, 3000], use a for loop to apply the discount rules from Task 1 to each order and print a summary table showing: order_amount -> discount% -> final_amount.
2. Also compute and print the total revenue after discounts.
3. Extra (optional): Print the number of orders that received a discount (discount > 0).

```python
# Task 2: Multiple Orders 

# 1. Given list of order amounts
orders = [1200, 2500, 800, 1750, 3000]

# to track totals
total_revenue = 0
discounted_orders_count = 0

# table header
print(f"{'Order Amount':<15} | {'Discount %':<12} | {'Final Amount':<15}")
print("-" * 48)

# applying discount rules from Task 1
for order_amount in orders:
    # Applying rules
    if order_amount >= 2000:
        discount_percent = 15
    elif 1500 <= order_amount < 2000:
        discount_percent = 10
    elif 1000 <= order_amount < 1500:
        discount_percent = 7
    else:
        discount_percent = 0

    # Calculate final amount
    discount_amount = (discount_percent / 100) * order_amount
    final_amount = order_amount - discount_amount
    
    # Update total revenue
    total_revenue += final_amount
    
    # Counting orders that received a discount
    if discount_percent > 0:
        discounted_orders_count += 1
        
    # summary row
    print(f"INR {order_amount: <11} | {discount_percent : <11}% | INR {final_amount:<11.2f}")

# 2. total revenue
print("-" * 48)
print(f"Total Revenue after discounts: INR {total_revenue:.2f}")

# output
print(f"Number of orders that received a discount: {discounted_orders_count}")
```

### Task 3: user Menu (while loop + break/continue)
1. Implement a simple while-loop driven menu that keeps asking the user to choose an action until they press q to quit.
Menu options:
1.1 -> Add order amount to a running list (use input())
,1.2 —> Show all orders and totals after applying discounts
,1.3 q —> Quit

2.  Use continue to re-show the menu after invalid input and break to exit on q.

```python
# Task 3: User Menu (while loop + break/continue)

# Initialize a running list to store order amounts
order_list = []

while True:
    # 1. Display Menu
    print("\n--- E-Commerce Inventory Menu ---")
    print("1 — Add order amount")
    print("2 — Show all orders and totals (with discounts)")
    print("q — Quit")
    
    choice = input("Select an option: ").strip().lower()

    # 2. Handle Quit option
    if choice == 'q':
        print("Exiting program. Goodbye!")
        break

    # 3. Add order amount
    if choice == '1':
        amount_input = input("Enter the order amount: ")
        
        # Check if input is numeric
        if amount_input.isdigit():
            order_list.append(int(amount_input))
            print(f"INT {amount_input} added successfully.")
        else:
            print("Error: Please enter a valid numeric integer.")
            continue # Re-show menu
            
    # 4.Showing all orders and totals
    elif choice == '2':
        if not order_list:
            print("No orders recorded yet.")
            continue
            
        print(f"\n{'Order #':<10} | {'Original':<12} | {'Discount %':<12} | {'Final Amount':<15}")
        print("-" * 55)
        
        running_total_revenue = 0
        order_count = 1
        
        for amount in order_list:
            if amount >= 2000:
                discount_percent = 15
            elif 1500 <= amount < 2000:
                discount_percent = 10
            elif 1000 <= amount < 1500:
                discount_percent = 7
            else:
                discount_percent = 0
            
            discount_value = (discount_percent / 100) * amount
            final_price = amount - discount_value
            running_total_revenue += final_price
            
            print(f"{order_count:<10} | INR {amount:<8} | {discount_percent:<11}% | INR {final_price:<11.2f}")
            order_count += 1
            
        print("-" * 55)
        print(f"Total Revenue: INR {running_total_revenue:.2f}")

    # 5. Handle invalid input
    else:
        print("Invalid selection. Please choose 1, 2, or q.")
        continue
```

### Task 4: Loop Control with Conditions (break & continue)
1. Given a list of daily sales: daily = [200, 150, 0, 400, 50, -1, 300], iterate through it with a for loop and:
- If a day's value is -1 , treat it as corrupted data and break the loop(stop processing).
- If a day's value is 0, treat it as a day with no sales and continue (skip adding to revenue).
- For valid positive sales, add to total_sales and print the running total.
2. Print the final total after the loop completes (or stops due to corrupted data).

```python
# Task 5: Loop Control with Conditions (break & continue)

# 1. Given list 
daily = [200, 150, 0, 400, 50, -1, 300]
total_sales = 0

print("Processing daily sales...")
print("-" * 30)

for sale in daily:
    # If a day's value is -1, treat it as corrupted data and break the loop
    if sale == -1:
        print("Corrupted data found (-1). Stopping all processing.")
        break
    
    # If a day's value is 0, treat it as no sales and continue (skip)
    if sale == 0:
        continue
    
    # For valid positive sales, add to total_sales and print running total
    total_sales += sale
    print(f"Sale recorded: INR {sale} | Running Total: INR {total_sales}")

# 2. Print the final total after the loop completes or breaks
print("-" * 30)
print(f"Final Total Sales: INR {total_sales}")
```
