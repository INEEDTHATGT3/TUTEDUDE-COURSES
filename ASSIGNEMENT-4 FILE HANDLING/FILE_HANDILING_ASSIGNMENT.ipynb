### Task 1: Write Sales Records to a File
1. Create a list of sales amounts:
2. sales = [1200, 450, 980, 1500, 3000]
3. Open a file named sales_data.txt in write mode.
4. Write each sale on a new line in the file.
5. Close the file, then reopen it and print its contents.
6. Extra (optional): Write the data in comma-separated format instead of separate lines.

```python
#1. Creating a list of sales amount
sales = [1200, 450, 980, 1500, 3000]

#2&3. Open sales_data.txt in W and writing each sale on new line
with open("sales_data.txt", "w") as file:
    for sale in sales:
        file.write(str(sale) + "\n")

#4. Close (handled by "with"), reopen, and printing content
print("Contents of sales_data.txt (New line format): ")
with open("sales_data.txt", "r") as file:
    content = file.read()
    print(content)

#Extra: write the data in csv
with open("sales_data_csv.txt", "w") as file:
    #converting all integers to strings and joining them with comma
    csv_string = ",".join([str(sale) for sale in sales])
    file.write(csv_string)

print("Contents of sales_data_csv.txt: ")
with open("sales_data_csv.txt", "r") as file:
    print(file.read())
```

### Task 2: Read File in Different ways
- Using the same sales_data.txt:
1. Read the entire file using .read() and print it.
2. Read the first line using .readline().
3. Read all lines using .readlines() and convert them into a list of integers.
- Ensure proper formatting and cleanup of newline characters.

```python
filename = "sales_data.txt"

#1. reading entire file
print("------- Method 1: --------")
with open(filename, "r") as file: 
    full_content = file.read()
    print(full_content)

# 2. Reading the first line using .readline()
print("------- Method 2: -------")
with open(filename, "r") as file:
    first_line = file.readline()
    # .strip() removes \n for clean printing
    print(f"First Line: {first_line.strip()}")

# 3. Reading all lines using .readlines() then converted them to integers
print("-------- Method 3: --------")
with open(filename, "r") as file:
    lines = file.readlines()
    
    # Using list comprehension to strip newlines and cast to int
    sales_list = [int(line.strip()) for line in lines]
    
    print(f"List of Integers: {sales_list}")
    print(f"Type of first element: {type(sales_list[0])}")

```

### Task 3: Append New Sales
1. Append these new sales to the same file: 5000, 2500, 1700
2. After appending, reopen and print the entire updated file.
3. Extra (optional): Print the total number of lines after appending.

```python
# listing out new sales
new_sales = [5000, 2500, 1700]
filename = "sales_data.txt"

# Appending new sales
with open(filename, "a") as file:
    for sale in new_sales:
        file.write(str(sale) + "\n")

# Reopening to read file
print("---- Updated sales_data.txt Content ----")
with open(filename, "r") as file:
    content = file.read()
    print(content)

# Extra: printing total no of lines after appending
with open(filename, "r") as file:
    lines = file.readlines()
    print(f"Total no. of sales records: {len(lines)}")
```

### Task 4: Generate Summary Report from File
- Using only file reading operations:
1. Read all sales values from sales data.txt.
2. Convert them into integers.
3. Calculate and print: 
- Total Sales
- Highest Sale
- Lowest Sale
- Average Sale
  Do not use any advanced libraries.

```python
filename = "sales_data.txt"
sales_list = []

#1. Read and convert to int
with open(filename, "r") as file:
    for line in file:
        clean_line = line.strip()
        if clean_line: # empty line edge case
            sales_list.append(int(clean_line))

#2. Caluclating statistics
if sales_list:
    total_sales = sum(sales_list)
    highest_sales = max(sales_list)
    lowest_sales = min(sales_list)
    average_sales = total_sales/ len(sales_list)

#3. printing summary
    print("--- SALES SUMMARY REPORT ---")
    print(f"Total Sales: {total_sales} INR")
    print(f"Highest Sale: {highest_sales} INR")
    print(f"Lowest Sale: {lowest_sales} INR")
    print(f"Average Sale: {average_sales:.2f} INR")
    print(f"Number of Records: {len(sales_list)}")
    print("----------------------------")
else:
    print("No data found in the file.")
```

### Task 5: Create Product Info File (User Input)
1. Ask the user for 3 product names & their prices.

4. Read the file and print each line with proper formatting.

```python
import sys #although follows a default behaviour exiting but just trying

filename = "products.txt"
product_entries = []

#1. Asking user for 3 products-> name, price
print("Enter Product Details: ")
for i in range(3):
    name = input(f" Product Name {i+1}: ")
    price_input = input(f"Enter Price For Product {i+1}: ")

    # Validating data
    if not price_input.isdigit():
        print("Input Invalid")
        sys.exit() # -> Execution termination

    # Storing format
    product_entries.append(f"{name} | {price_input}")

#2. Writing to products.txt
with open(filename, "w") as file:
    for entry in product_entries:
        file.write(entry + "\n")

#3. Read and Print 
print(f" \n -------  PRODUCT REPORT -----------")
print(f"{'Product Name': <10} \t|\t {'Price':<10} ")
with open(filename, "r") as file:
    for line in file:
        #Error 1: did used file intead of line ->fixed (i just looked at heading appearing as output and moved to next task :my mistake )
        if "|" in line:
            #splitting line into components
            parts = line.split("|")
            p_name = parts[0].strip()
            p_price = parts[1].strip()
                #Printing
            #Error 2: p_name,p_price used as string rather then object -> fixed  
            print(f"{p_name:<10} \t|\t{p_price:<10} INR")
print("-" * 40)



```

### Task 7: Mini Project - Export Discounted Prices
- Create a dictionary:
prices = {
"Mouse": 500,
"Keyboard": 800,
"Monitor": 7000,
"Pendrive": 400,
"Camera": 5000 }
1. Ask the user for a discount percentage.
2. Write discounted prices into discount_report.txt using:
a. Product I Original Price I Discounted Price
,b.After writing, read the file and print it to the terminal.
3. Extra (optional): Write a summary at the bottom of the file:
 - Total Items: X
 - Average Discounted Price: Y

```python
import sys

# 1. Data intialization
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}

filename = "discount_report.txt"

# Getting Input and validating
# Thank you sir for pointing out somehow my IDE and python didn't throw error in my setup environment
try:
    discount_input = input("Enter discount % (0-100): ")
    discount= float(discount_input)

    # Within Range 0-100
    if not(0 <= discount <= 100):
        raise ValueError("Discount must be between 0-100")
    
except ValueError:
    print("Input Invalid: non numeric")
    sys.exit()
''' if not discount_input.replace('.', '').isdigit():
    print("input invalid")
    sys.exit() '''

discount_percent = float(discount_input)

# 3. Processing Data and Write to File
discounted_prices_list = []

with open(filename, "w") as file:
    file.write(f"{'Product':<10} \t| {'Original Price':<10} \t| {'Discounted Price':<10}\n")
    file.write("-" * 45 + "\n")
    
    for product, original_price in prices.items():
        # Discounted Price Calculation
        discount_amount = (discount_percent / 100) * original_price
        discounted_price = original_price - discount_amount
        discounted_prices_list.append(discounted_price)
        
        # formatted row written in file
        file.write(f"{product:<10}\t|\t {original_price:<10}\t|\t {discounted_price:.2f}\n")
    
    # 4. Extra: Writing Summary at the bottom
    total_items = len(prices)
    avg_discounted = sum(discounted_prices_list) / total_items
    
    file.write("-" * 45 + "\n")
    file.write(f"Total Items: {total_items}\n")
    file.write(f"Average Discounted Price: {avg_discounted:.2f} INR\n")

# 5. Reading the file 
print("\n--- GENERATED REPORT ---")
with open(filename, "r") as file:
    print(file.read())
```
