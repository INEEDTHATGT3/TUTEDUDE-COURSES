### Task 3: Product Pricing (Dictionaries)
1. Create a dictionary price_dict where keys are product names and values are prices (integers or floats). Include at least 6 entries.
2. Write small code blocks to:
Add a new product with price to price_dict.
Update the price of an existing product.
Remove a product by name (handle the case when the product does not exist).
3. Print the average price of all products (use only dictionary operations and basic arithmetic).
4. Extra (optional): Print the product with both the maximum and minimum prices.

```python
# Task 3: Product Pricing (Dictionaries)

# 1. Create a dictionary price_dict where keys are product names and values are prices
price_dict = {
    "Laptop": 1200,
    "Smartphone": 800,
    "Wireless Headphones": 150,
    "Mechanical Keyboard": 100,
    "Gaming Mouse": 60,
    "External SSD": 120
}

# 2. Write small code blocks for dictionary operations

# Add a new product with price to price_dict
price_dict["Webcam"] = 55
print("Added Webcam. Updated dictionary:", price_dict)

# Update the price of an existing product
price_dict["Smartphone"] = 750
print("Updated Smartphone price to USD 750.")

# Remove a product by name (handle the case when the product does not exist)
product_to_remove = "Gaming Mouse"
if product_to_remove in price_dict:
    del price_dict[product_to_remove]
    print(f"Removed {product_to_remove} from inventory.")
else:
    print(f"Error: {product_to_remove} not found in inventory.")

# 3. Print the average price of all products
# Using sum() on values and len() for the count
total_sum = sum(price_dict.values())
count = len(price_dict)
average_price = total_sum / count
print(f"The average price of all products is: USD {average_price:.2f}")

# 4. Extra (optional): Print the product with both the maximum and minimum prices
max_product = max(price_dict, key=price_dict.get)
min_product = min(price_dict, key=price_dict.get)

print(f"Product with maximum price: {max_product} (USD {price_dict[max_product]})")
print(f"Product with minimum price: {min_product} (USD {price_dict[min_product]})")
```

### Task 4: Combined Operations
1. Using the products list and price_dict, create a list of tuples named catalog where each tuple is (product_name, price, category).
2. From catalog, create a new dictionary category_to_products that maps each category to a list of product names in that category.
3. Print all products that belong to the category that has the maximum number of Products.

```python
# Task 4: Combined Operations

# Setup: Data from previous tasks
products = ["Laptop", "Smartphone", "Wireless Headphones", "Mechanical Keyboard", "External SSD", "Webcam"]
categories_parallel = ["Electronics", "Electronics", "Accessories", "Peripherals", "Storage", "Accessories"]
price_dict = {
    "Laptop": 1200,
    "Smartphone": 750,
    "Wireless Headphones": 150,
    "Mechanical Keyboard": 100,
    "External SSD": 120,
    "Webcam": 55
}

# 1. Using the products list and price_dict, create a list of tuples named catalog 
# Each tuple is (product_name, price, category)
catalog = []
for i in range(len(products)):
    name = products[i]
    price = price_dict[name]
    category = categories_parallel[i]
    catalog.append((name, price, category))

print("Catalog of products:", catalog)

# 2. Create a new dictionary category_to_products mapping each category to a list of product names
category_to_products = {}
for name, price, category in catalog:
    if category not in category_to_products:
        category_to_products[category] = []
    category_to_products[category].append(name)

print("\nCategory to Products Mapping:", category_to_products)

# 3. Print all products that belong to the category that has the maximum number of products
max_category = ""
max_count = 0

for category, product_list in category_to_products.items():
    if len(product_list) > max_count:
        max_count = len(product_list)
        max_category = category

print(f"\nCategory with the most products: {max_category}")
print(f"Products in {max_category}:", category_to_products[max_category])
```
