## Task 7 - Mini Problem: Menu Using Functions
- Create the following three small functions:
1. add_price(prices_list, price) —+ adds price to the list.
2. get_average_price(prices list) —+ returns average price.
3. get_max_price(prices_list) -+ returns the maximum price.
- Then create a simple menu:
1. —+ Add price
2. —+ Show average price
3. —+ Show highest price
4. q —+ Quit
- Use only loops + function calls (no OOP).

```python
def add_price(prices_list, price):
    prices_list.append(price)

def get_average_prices(prices_list):
    if not prices_list:
        return 0
    return sum(prices_list)/ len(prices_list)

def get_max_prices(prices_list):
    if not prices_list:
        return 0
    return max(prices_list)

inventory_prices = []

while True:
    print("\n Biling Utility Menu")
    print("1 -> Add price")
    print("2 -> Show average price")
    print("3 -> Show highest price")
    print("q -> Quit")

    choice = input("Select an opton: ").strip().lower()

    if choice == '1':
        raw_input = input("Enter the price to add: ")
        #input validation without creating exception
        if raw_input.replace('.', '', 1).isdigit():
            new_price = float(raw_input)
            add_price(inventory_prices, new_price)
            print(f"Price INR {new_price:.2f} added successfully.")
        else:
            print("invalid input. Please enter a valid number")
    
    elif choice == '2':
        avg = get_average_prices(inventory_prices)
        print(f"Average Price: INR {avg:.2f}")

    elif choice == '3':
        highest = get_max_prices(inventory_prices)
        print(f"Highest Price: INR {highest:.2f}")

    elif choice == 'q':
        print("Exiting billing utility. Goodbye!")
        break

    else:
        print("Invalid choice. Please select 1, 2, 3, or q.")
```
