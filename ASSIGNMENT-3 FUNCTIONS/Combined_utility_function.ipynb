## Task 6 - Combined Utility Function
- Write a function process_prices(prices) that:
1. Takes a list of prices.
2. Uses map() + lambda to apply 10% discount to all prices.
3. Uses filter() to keep only discounted prices that are above 300.
4. Returns both lists:
- discounted_prices
- filtered_prices
- Test with: process_prices[500, 900, 50, 750]

```python
def process_prices(prices):
    # Exception handiling
    if not prices:
        return "No prices provided"
    # 1. Applying 10% discount and rounding to 2% decimal places
    discounted_prices = list(map(lambda p: round(p * 0.9, 2), prices))

    #2. Filtering  out prices
    filtered_prices = list(filter(lambda p: p>300, discounted_prices))

    return{
        "discounted_prices" : discounted_prices,
        "filtered prices" : filtered_prices,
    }

# Testing the function
test_prices = [500, 900, 50, 750]
results = process_prices(test_prices)

print(f"Orignal Input: {test_prices}")
print(f"All dicounted prices: {results['discounted_prices']}")
print(f"Filtered prices (>300): {results['filtered prices']}")

process_prices([])
```
