## Task 3 - Lambda Function: GST Calculator
- Create a lambda function gst that returns price after adding 18% GST.
Example:
1. gst = lambda price: price + (0.18 * price)
2. print(gst(100)) # should return 118
- Extra (optional): Create another lambda to compute final price after GST + discount together.

```python
# Lambda for 18% GST
gst = lambda price: price + (0.18 * price)

# Extra: lambda for Final price After GST and Discount
final_price_calc = lambda price, discount: gst(price) * (1-discount/100)

# Testing Functions
base_price = 1000
discount_val = 10

# Testing
print(f"Price after GST: {gst(base_price): .2f}")
print(f"Final Price after GST and {discount_val}% Discount: {final_price_calc(base_price, discount_val): .2f}")
```
