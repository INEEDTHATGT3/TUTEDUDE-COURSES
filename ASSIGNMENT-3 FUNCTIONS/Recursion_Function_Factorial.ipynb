## Task 2 - Recursive Function: Factorial Utility
- Create a recursive function factorial(n) that:
1. Returns the factorial of n.
2. Handles two edge cases: n == 0 and n == 1.
3. Prints an error message if n is negative.
- Test with:
1. factorial(5)
2. factorial(0)
3. factorial(-3)

```python
def factorial(n):
    # invalid input
    if n<0:
        print(f"Invalid input: {n}, is negative")
        return None     
    # base case
    if n == 0 or n == 1:
       return 1
        
    # recursive case
    return n * factorial(n-1)

# Testing Function
print(f"Factorial(5): {factorial(5)}")
print(f"Factorial(0): {factorial(0)}")
print(f"Factorial(-3): {factorial(-3)}")
```
